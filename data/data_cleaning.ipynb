## Data Cleaning

```python
#filter out undesired question types
clean_df = df[(df['question_type']=='Default question type') |
              (df['question_type']=='Other common hallucinated questions')]
clean_df.head()
```

```python
#omit irrelevant columns
clean_df = clean_df[['reference', 'input', 'output','label', 'hallucination_type_realized', 'question_type', 'hallucination_type_encouraged']]
clean_df.head()
```

```python
#drop rows with missing values in output column
clean_df = clean_df[clean_df['output'].notna()]
```

```python
#check missing values
clean_df.isnull().sum()
```

```python
clean_df.shape
```

```python
clean_df.to_csv('cleaned_data.csv', index=False)
```
